##L09: Unity Catalog: names, comments, tags, grants, lineage
###Goal
Explore governance features in Unity Catalog and document your tables so people (and Genie) can use them.

###Step 1. Explore the hierarchy

```python
%sql
SHOW CATALOGS;
SHOW SCHEMAS IN da;
SHOW TABLES IN da.silver;
SHOW VOLUMES IN da.bronze;
DESCRIBE EXTENDED da.silver.yellow_trips;
```

###Step 2. Add comments and tags
Good metadata is how tables become discoverable.

```python
%sql
COMMENT ON TABLE da.silver.yellow_trips IS 'Cleaned NYC yellow taxi trips that passed all quality rules. Source: NYC TLC trip records.';
ALTER TABLE da.silver.yellow_trips ALTER COLUMN trip_id COMMENT 'SHA-256 hash of vendor, pickup, dropoff, zones and amounts';
ALTER TABLE da.silver.yellow_trips ALTER COLUMN tip_amount COMMENT 'Tip in USD. Cash tips are not recorded and appear as 0';
ALTER TABLE da.silver.yellow_trips SET TAGS ('layer' = 'silver', 'source' = 'nyc_tlc');
```

###Step 3. Look at privileges
You are the owner, so you can do everything. Practise one grant and revoke, then check the result.

```python
%sql
SHOW GRANTS ON TABLE da.silver.yellow_trips;
SHOW GRANTS ON SCHEMA da.silver;
```

```python
%sql
GRANT SELECT ON TABLE da.silver.yellow_trips TO `account users`;
SHOW GRANTS ON TABLE da.silver.yellow_trips;
REVOKE SELECT ON TABLE da.silver.yellow_trips FROM `account users`;
```

###Step 4. Information schema

```python
%sql
SELECT table_schema, table_name, table_type FROM da.information_schema.tables
WHERE table_schema IN ('bronze','silver','gold') ORDER BY 1,2;
SELECT table_name, column_name, comment FROM da.information_schema.columns
WHERE table_schema = 'silver' AND comment IS NOT NULL;
```
